```python
import os
import joblib
import numpy as np
import pandas as pd
import shap
import matplotlib.pyplot as plt
import seaborn as sns
from collections import defaultdict
from sklearn.preprocessing import OneHotEncoder

# === Setup paths ===
model_base_path = "/n/groups/patel/adithya/Alz_Outputs/Final_Outputs/Multirun_cell_on_cell_both_new"
expr_path = "/home/adm808/NormalizedCellMatrixSyn18485175.parquet"
metadata_path = "/home/adm808/New_CellMetadataSyn1848517.parquet"
output_dir = "/n/groups/patel/adithya/Alz_Outputs/Final_Outputs/Final_Outputs_Figures/AD_prediction_stuff_new/Figure9/SHAP_by_APOE/Final/"
os.makedirs(output_dir, exist_ok=True)

# === Cell types ===
cell_types = {
    "Ast": "Astrocytes",
    "Ex": "Excitatory Neurons",
    "In": "Inhibitory Neurons",
    "Mic": "Microglia",
    "Oli": "Oligodendrocytes",
    "Opc": "OPCs"
}

# === APOE genotype groups ===
def map_apoe_group(row):
    if row.get("apoe_genotype_440", 0) == 1:
        return "ε4/4"
    elif row.get("apoe_genotype_340", 0) == 1:
        return "ε3/4"
    else:
        return "Not ε4 carrier"

# === Load expression and metadata ===
expr = pd.read_parquet(expr_path).T  # rows = TAGs
expr.columns = expr.columns.str.replace(r'[^a-zA-Z0-9]', '', regex=True)
metadata = pd.read_parquet(metadata_path)

# Preprocess age
metadata.loc[metadata.age_death == '90+', 'age_death'] = 90
metadata['age_death'] = pd.to_numeric(metadata['age_death'], errors='coerce')

# One-hot encode APOE genotype
metadata = pd.get_dummies(metadata, columns=["apoe_genotype"])


rename_dict = {
    "apoe_genotype_23.0": "apoe_genotype_230",
    "apoe_genotype_24.0": "apoe_genotype_240",
    "apoe_genotype_33.0": "apoe_genotype_330",
    "apoe_genotype_34.0": "apoe_genotype_340",
    "apoe_genotype_44.0": "apoe_genotype_440",
}
metadata = metadata.rename(columns=rename_dict)

apoe_columns = [col for col in metadata.columns if col.startswith("apoe_genotype")]

# Merge expression + metadata
metadata = metadata.set_index('TAG')
common_tags = metadata.index.intersection(expr.index)
X_full = pd.concat([expr.loc[common_tags], metadata.loc[common_tags]], axis=1)

cell_type_shap_values = {}

# === Main loop ===
for short_label, pretty_label in cell_types.items():
    print(f"Processing: {pretty_label}")

    # shap_values_all = defaultdict(list)
    shap_values_all = defaultdict(list)
    group_membership_all = []

    for split in range(1, 6):
        model_path = os.path.join(model_base_path, short_label, f"split_{split}", "maximal_classifier.joblib")
        if not os.path.exists(model_path):
            print(f"Model missing: {model_path}")
            continue

        # model = joblib.load(model_path)
        # model.model.booster_
        # automl_model = joblib.load(model_path)
        # lgbm_booster = automl_model.model.booster_
        # features = model.feature_names_in_

        automl_model = joblib.load(model_path)
        lgbm_booster = automl_model.model.model.booster_
        features = automl_model.feature_names_in_
        meta_feats = ['educ','age_death','msex'] + apoe_columns
        gene_feats = [f for f in features if f not in meta_feats]

        X = X_full.copy()
        available_features = [f for f in features if f in X.columns]
        X = X[features].copy()
        # X = X[available_features].copy()
        # X = X[features].copy()
        X = X.fillna(0)  # safety

        pandas_cats = lgbm_booster.pandas_categorical  # list of categories per feature
        for feat, cats in zip(features, pandas_cats):
            if cats:
                # force the same categorical dtype
                X[feat] = pd.Categorical(X[feat], categories=cats)
            else:
                # numeric feature
                X[feat] = pd.to_numeric(X[feat], errors="coerce")

        # Add APOE group label
        # X['APOE_GROUP'] = metadata.loc[X.index].apply(map_apoe_group, axis=1)

        # #explainer = shap.TreeExplainer(model)
        # print("Reached explainer")
        # explainer = shap.TreeExplainer(lgbm_booster)
        # shap_vals = explainer.shap_values(X.drop(columns=["APOE_GROUP"], errors='ignore'))

        # # SHAP returns a list of arrays for multiclass — assume binary classification
        # if isinstance(shap_vals, list):
        #     shap_vals = shap_vals[1]

        # shap_df = pd.DataFrame(shap_vals, columns=features, index=X.index)
        # shap_df["APOE_GROUP"] = X["APOE_GROUP"]
        # group_membership_all.append(shap_df["APOE_GROUP"])
        # shap_df = shap_df.drop(columns=["APOE_GROUP"])

        # for group in ["ε3/3", "ε3/4 or ε2/4", "ε4/4"]:
        #     print(f"[DEBUG] Unique APOE_GROUPs for this split: {group_membership_all[-1].unique()}")
        #     group_cells = shap_df.index[group_membership_all[-1] == group]
        #     if len(group_cells) < 10:
        #         continue
        #     group_mean = shap_df.loc[group_cells].abs().mean()
        #     shap_values_all[group].append(group_mean)
        # Add APOE group label
        X['APOE_GROUP'] = metadata.loc[X.index].apply(map_apoe_group, axis=1)

        for group in ["Not ε4 carrier", "ε3/4", "ε4/4"]:
            group_cells = X.index[X['APOE_GROUP'] == group]
            if len(group_cells) < 10:
                print(f"Skipping group {group} — too few cells.")
                continue

            # Restrict to cells of this group only
            X_group = X.loc[group_cells].drop(columns=["APOE_GROUP"])
            
            
            # Compute SHAP for only this group
            print(f"Computing SHAP for group {group} with {len(group_cells)} cells.")
            explainer = shap.TreeExplainer(lgbm_booster)
            shap_vals = explainer.shap_values(X_group)

            if isinstance(shap_vals, list):  # binary case
                shap_vals = shap_vals[1]

            shap_df = pd.DataFrame(shap_vals, columns=features, index=X_group.index)
            group_mean = shap_df.abs().mean()

            shap_values_all[group].append(group_mean)


            # ➡️ 1) cell‐level for gene SHAPs
            cell_gene_mean = shap_df[gene_feats].abs().mean()
            shap_values_all[f"{group}_genes_cell"].append(cell_gene_mean)

            # ➡️ 2) donor‐level for metadata SHAPs
            donor_series      = metadata.loc[shap_df.index, 'sample']
            donor_meta_mean   = (
                shap_df[meta_feats]
                .assign(donor=donor_series)
                .groupby('donor')
                .mean()
            )
            meta_group_mean   = donor_meta_mean.abs().mean()
            shap_values_all[f"{group}_meta_donor"].append(meta_group_mean)
        # define once above the loop
        

        # for group in ["ε3/3", "ε3/4 or ε2/4", "ε4/4"]:
        #     mask     = (group_membership_all[-1] == group)
        #     group_df = shap_df.loc[mask]
        #     if group_df.shape[0] < 10:
        #         continue

        #     # ➡️ 1) cell‐level for gene SHAPs
        #     cell_gene_mean = group_df[gene_feats].abs().mean()
        #     shap_values_all[f"{group}_genes_cell"].append(cell_gene_mean)

        #     # ➡️ 2) donor‐level for metadata SHAPs
        #     donor_series      = metadata.loc[group_df.index, 'sample']
        #     donor_meta_mean   = (
        #         group_df[meta_feats]
        #         .assign(donor=donor_series)
        #         .groupby('donor')
        #         .mean()
        #     )
        #     meta_group_mean   = donor_meta_mean.abs().mean()
        #     shap_values_all[f"{group}_meta_donor"].append(meta_group_mean)

    # === Plotting ===
    for group in ["Not ε4 carrier", "ε3/4", "ε4/4"]:
        if group not in shap_values_all:
            print(f"No data for group {group} in {pretty_label}")
            continue

        
        # avg_shap = pd.concat(shap_values_all[group], axis=1).mean(axis=1)
        # avg_shap = avg_shap[avg_shap > 0]

        # Compute mean and variance across splits
        shap_concat = pd.concat(shap_values_all[group], axis=1)
        # Collect all features across all splits
        all_features = set()
        for df in shap_values_all[group]:
            all_features.update(df.index)

        # Build a matrix with NaNs where features are missing
        shap_matrix = pd.DataFrame(index=list(all_features))
        # shap_matrix = pd.DataFrame(index=all_features)

        for i, df in enumerate(shap_values_all[group]):
            shap_matrix[f"split_{i}"] = df

        # Fill missing values with 0s and compute corrected mean and variance
        shap_matrix = shap_matrix.fillna(0)
        avg_shap = shap_matrix.mean(axis=1)
        var_shap = shap_matrix.var(axis=1)

        # Only keep features with:
        # 1. Positive mean SHAP → AD drivers
        # 2. Non-zero variance → varies in group
        avg_shap = avg_shap[(avg_shap > 0) & (var_shap > 0)]
        # Rename to human-readable names
        pretty_names = {
            # "age_death": "Age of Death",
            # "educ": "Education",
            # "msex": "Sex",
            "apoe_genotype_230": "APOE ε2/3",
            "apoe_genotype_330": "APOE ε3/3",
            "apoe_genotype_340": "APOE ε3/4",
            "apoe_genotype_440": "APOE ε4/4",
        }
        avg_shap = avg_shap.rename(pretty_names)
        top_features = avg_shap.nlargest(30)

        # plt.figure(figsize=(8, 6))
        # sns.barplot(x=np.sqrt(top_features.values), y=top_features.index, color="steelblue")
        # # sns.barplot(x=top_features.values, y=top_features.index, color="steelblue")
        # plt.title(f"Top Predictive Features in {pretty_label} ({group})")
        # # plt.xlabel("Mean |SHAP value|")
        # plt.xlabel("√(Mean |SHAP value|)")
        # plt.ylabel("Feature Name")
        # plt.tight_layout()

        # save_file = os.path.join(output_dir, f"{short_label}_top30_{group.replace('/', '').replace(' ', '_')}.png")
        # #plt.savefig(save_file, dpi=300)
        # #plt.show()

        # print(f"Saved: {save_file}")

    # === Plotting metadata‐level SHAP (per donor) ===
    meta_groups = ["Not ε4 carrier", "ε3/4", "ε4/4"]
    for group in meta_groups:
        key = f"{group}_meta_donor"
        if key not in shap_values_all or not shap_values_all[key]:
            print(f"No metadata‐level data for group {group} in {pretty_label}")
            continue

        avg_meta = pd.concat(shap_values_all[key], axis=1).mean(axis=1)
        avg_meta = avg_meta.rename(pretty_names).sort_values(ascending=False)

        plt.figure(figsize=(6, 4))
        # sns.barplot(x=np.sqrt(top_features.values), y=top_features.index, color="steelblue")
        sns.barplot(x=np.sqrt(avg_meta.values), y=avg_meta.index, color="salmon")
        plt.title(f"Metadata SHAP in {pretty_label} ({group})")
        plt.xlabel("√(Mean |SHAP value|) (per donor)")
        #plt.xlabel("√(Mean |SHAP value|)")
        plt.ylabel("Feature Name")
        plt.tight_layout()

        save_file = os.path.join(
            output_dir,
            f"{short_label}_meta_shap_{group.replace('/', '').replace(' ', '_')}.png"
        )
        plt.savefig(save_file, dpi=300)
        plt.show()
        print(f"Saved metadata plot: {save_file}")

    
    # === Plotting gene‐level SHAP (per cell) ===
    for group in meta_groups:
        key = f"{group}_genes_cell"
        if key not in shap_values_all or not shap_values_all[key]:
            print(f"No gene‐level data for group {group} in {pretty_label}")
            continue

        avg_shap = pd.concat(shap_values_all[key], axis=1).mean(axis=1)
        top_genes = avg_shap[avg_shap > 0].nlargest(20)
        # top_genes = avg_shap.nlargest(20)

        plt.figure(figsize=(8, 6))
        sns.barplot(x=np.sqrt(top_genes.values), y=top_genes.index, color="steelblue")
        plt.title(f"Top Genes driving {pretty_label} toward AD ({group})")
        plt.xlabel("√(Mean |SHAP value|) (per cell)")
        plt.ylabel("Feature Name")
        plt.tight_layout()

        

        save_file = os.path.join(
            output_dir,
            f"{short_label}_top20_genes_{group.replace('/', '').replace(' ', '_')}.png"
        )
        plt.savefig(save_file, dpi=300)
        plt.show()
        print(f"Saved gene plot: {save_file}")

    cell_type_shap_values[short_label] = shap_values_all

import xlsxwriter
```

```python
# === Save SHAP data summary to Excel ===
output_dir = '/n/groups/patel/adithya/Alz_Outputs/Final_Outputs/Data/'
excel_path = os.path.join(output_dir, "SHAP_Summary.xlsx")
with pd.ExcelWriter(excel_path, engine="xlsxwriter") as writer:
    description = {
        "Sheet": [],
        "Description": []
    }

    for short_label, pretty_label in cell_types.items():
        shap_values_all = cell_type_shap_values[short_label] 

        # --- Metadata-level SHAP ---
        meta_df = pd.DataFrame()
        for group in ["Not ε4 carrier", "ε3/4", "ε4/4"]:
            key = f"{group}_meta_donor"
            if key in shap_values_all:
                group_df = pd.concat(shap_values_all[key], axis=1)
                split_cols = [f"Split{i+1}" for i in range(group_df.shape[1])]
                group_df.columns = split_cols
                group_df["Feature"] = group_df.index
                group_df["APOE Group"] = group
                group_df["Mean"] = group_df[split_cols].mean(axis=1)
                group_df["Std"] = group_df[split_cols].std(axis=1)
                group_df["SqrtMean"] = np.sqrt(group_df["Mean"])

                meta_df = pd.concat([meta_df, group_df], axis=0)
        
        if not meta_df.empty:
            meta_df = meta_df[["APOE Group", "Feature"] + [c for c in meta_df.columns if c.startswith("Split")] + ["Mean", "Std", "SqrtMean"]]
            sheetname = f"{short_label}_Meta"
            meta_df.to_excel(writer, sheet_name=sheetname, index=False)
            description["Sheet"].append(sheetname)
            description["Description"].append(f"Donor-level SHAP values (per split) for {pretty_label}")

        # --- Gene-level SHAP ---
        gene_df = pd.DataFrame()
        for group in ["Not ε4 carrier", "ε3/4", "ε4/4"]:
            key = f"{group}_genes_cell"
            if key in shap_values_all:
                group_df = pd.concat(shap_values_all[key], axis=1)
                split_cols = [f"Split{i+1}" for i in range(group_df.shape[1])]
                group_df.columns = split_cols
                group_df["Feature"] = group_df.index
                group_df["APOE Group"] = group
                group_df["Mean"] = group_df[split_cols].mean(axis=1)
                group_df["Std"] = group_df[split_cols].std(axis=1)
                group_df["SqrtMean"] = np.sqrt(group_df["Mean"])

                gene_df = pd.concat([gene_df, group_df], axis=0)
        
        if not gene_df.empty:
            gene_df = gene_df[["APOE Group", "Feature"] + [c for c in gene_df.columns if c.startswith("Split")] + ["Mean", "Std", "SqrtMean"]]
            sheetname = f"{short_label}_Gene"
            gene_df.to_excel(writer, sheet_name=sheetname, index=False)
            description["Sheet"].append(sheetname)
            description["Description"].append(f"Cell-level SHAP values (per split) for {pretty_label}")

    # --- Add README Sheet ---
    desc_df = pd.DataFrame(description)
    desc_df.to_excel(writer, sheet_name="README", index=False)

print(f" SHAP summary Excel workbook saved to: {excel_path}")
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from pathlib import Path
from scipy.stats import spearmanr

# === Setup ===
base_dir = Path("/n/groups/patel/adithya/Alz_Outputs/Final_Outputs/Multirun_cell_on_cell_genes_new")
expr_path = "/home/adm808/NormalizedCellMatrixSyn18485175.parquet"
metadata_path = "/home/adm808/New_CellMetadataSyn1848517.parquet"

genes_of_interest = ["RASGEF1B", "LINGO1", "APOE", "ARL17B", "APOD", "MT2A"]
cell_types = {
    "Ast": "Astrocytes",
    "Ex": "Excitatory Neurons",
    "In": "Inhibitory Neurons",
    "Mic": "Microglia",
    "Oli": "Oligodendrocytes",
    "Opc": "OPCs"
}
remove_zero_expression = True

# === Load ===
expr_df = pd.read_parquet(expr_path).T
expr_df.columns = expr_df.columns.str.replace(r'[^a-zA-Z0-9]', '', regex=True)
metadata_df = pd.read_parquet(metadata_path).set_index("TAG")

# === Matrix to store correlations ===
cor_matrix = pd.DataFrame(index=genes_of_interest, columns=cell_types.keys(), dtype=float)

for gene in genes_of_interest:
    for ct_key, ct_name in cell_types.items():
        rho_vals = []

        for split in range(1, 6):
            pred_path = base_dir / ct_key / f"split_{split}" / "test_predictions.csv"
            if not pred_path.exists():
                continue

            preds = pd.read_csv(pred_path)
            tags = preds["TAG"]
            if gene not in expr_df.columns:
                continue

            gene_expr = expr_df.loc[tags, gene]
            if remove_zero_expression:
                gene_expr = gene_expr[gene_expr > 0]
            valid_tags = gene_expr.index

            if len(valid_tags) < 5:
                continue

            y_pred = preds.set_index("TAG").loc[valid_tags]["predicted_proba"]
            rho, _ = spearmanr(gene_expr.loc[valid_tags], y_pred)
            rho_vals.append(rho)

        if len(rho_vals) > 0:
            cor_matrix.loc[gene, ct_key] = np.mean(rho_vals)
            # cor_matrix.loc[gene, ct_name] = np.mean(rho_vals)

# === Plot heatmap ===
plt.figure(figsize=(12, 6))
sns.heatmap(cor_matrix, annot=True, fmt=".2f", cmap="vlag", center=0, linewidths=0.5, cbar_kws={"label": "Spearman ρ"})
plt.title("Correlation Between Gene Expression and Predicted AD Probability")
plt.ylabel("Gene")
plt.xlabel("Cell Type")
plt.tight_layout()
plt.savefig("/n/groups/patel/adithya/Alz_Outputs/Final_Outputs/Final_Outputs_Figures/AD_prediction_stuff_new/Figure9/gene_celltype_spearman_heatmap.png", dpi=300)
plt.show()
```
