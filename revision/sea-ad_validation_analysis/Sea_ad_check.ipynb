```python
import scanpy as sc
import pandas as pd

file_path = '/n/scratch/users/a/adm808/SEA_AD/SEAAD_A9_RNAseq_final-nuclei.2024-02-13.h5ad'

# Load metadata only — no expression matrix in memory
adata = sc.read_h5ad(file_path, backed='r')
obs_df = adata.obs.copy()

# --- Diagnosis candidate columns ---
diag_candidates = [
    'Cognitive Status',
    'Overall AD neuropathological Change',
    'Severely Affected Donor'
]

for col in diag_candidates:
    print(f"\n{'='*50}")
    print(f"Column: {col}")
    print(f"Dtype:  {obs_df[col].dtype}")
    print(f"N unique: {obs_df[col].nunique()}")
    print(f"Value counts (cell-level):")
    print(obs_df[col].value_counts(dropna=False))
    print(f"\nUnique donor-level counts:")
    donor_counts = obs_df.groupby('Donor ID')[col].first().value_counts(dropna=False)
    print(donor_counts)
```

```python
# --- Cell type candidate columns ---
celltype_candidates = ['Class', 'Subclass', 'Supertype']

for col in celltype_candidates:
    print(f"\n{'='*50}")
    print(f"Column: {col}")
    print(f"N unique: {obs_df[col].nunique()}")
    print(f"Value counts:")
    print(obs_df[col].value_counts(dropna=False))

# Also print Subclass breakdown within each Class
# so you can see if the granularity matches Morabito's broad.cell.type
print(f"\n{'='*50}")
print("Subclass grouped by Class:")
print(obs_df.groupby(['Class', 'Subclass']).size().reset_index(name='cell_count').to_string(index=False))
```
